```python
import seaborn as sns 
sns.histplot(data=train_df,x='Sales',bins=100,kde=True)
```

```python
is_open=train_df[train_df['Open']==1]
sns.histplot(data=is_open,x='Sales',bins=100,kde=True)
```

```python
sns.histplot(data=train_df,x='Customers',bins=40)
```

```python
fig,axes=plt.subplots(1,2,figsize=(12,4))
sns.countplot(data=train_df,x='Promo',ax=axes[0])
sns.countplot(data=train_df,x='SchoolHoliday',ax=axes[1])
```

```python
fig1,axes1=plt.subplots(2,2,figsize=(12,6))
sns.histplot(data=store_df,x='CompetitionDistance',ax=axes1[0,0],bins=50)
sns.boxplot(data=store_df,y='CompetitionDistance',ax=axes1[0,1])
sns.countplot(data=store_df,x='StoreType',ax=axes1[1,0])
sns.countplot(data=store_df,x='Assortment',ax=axes1[1,1])

# we can see there are so many outliers in the dataset 
```

```python
# bivarient analysis
sns.scatterplot(data=train_df,x='Customers',y='Sales',hue='Promo')
```

```python
sns.barplot(data=train_df,x='DayOfWeek',y='Customers')
```

```python
sns.barplot(data=train_df,x='DayOfWeek',y='Sales')
```

```python
merge=pd.merge(train_df,store_df,on='Store',how='left')
```

```python
sns.scatterplot(data=merge,x='CompetitionDistance',y='Sales')
```

```python
fig,axes=plt.subplots(2,2,figsize=(20,20))
sns.boxplot(data=merge,x='Promo',y='Sales',ax=axes[0,0])
sns.barplot(data=merge,x='DayOfWeek',y="Sales",ax=axes[0,1])
sns.barplot(data=merge,x='StoreType',y="Sales",ax=axes[1,0])
sns.barplot(data=merge,x='Assortment',y="Sales",ax=axes[1,1])
```
